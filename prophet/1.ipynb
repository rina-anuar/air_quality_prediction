```python
import requests
import pandas as pd

# Координаты Алматы и твои даты
url = "https://archive-api.open-meteo.com/v1/archive"
params = {
    "latitude": 43.2389,
    "longitude": 76.8897,
    "start_date": "2020-02-10",
    "end_date": "2024-08-12",
    "daily": ["temperature_2m_mean", "wind_speed_10m_max"],
    "timezone": "Asia/Almaty"
}

response = requests.get(url, params=params)
data = response.json()

# Превращаем в таблицу
weather_df = pd.DataFrame(data["daily"])
#weather_df.columns = ['ds', 'temp', 'wind_speed']
weather_df
```
